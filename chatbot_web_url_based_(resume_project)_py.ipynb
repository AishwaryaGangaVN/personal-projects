```python
import requests
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.document_loaders import UnstructuredURLLoader
from langchain.vectorstores import FAISS
from sentence_transformers import SentenceTransformer
import faiss
import numpy as np

# Function to validate URL accessibility
def is_valid_url(url):
    try:
        response = requests.get(url)
        return response.status_code == 200
    except Exception as e:
        print(f"Error accessing {url}: {e}")
        return False

# Load the embedding model

from langchain.embeddings import HuggingFaceEmbeddings

# Load the embedding model
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")




# List of URLs
urls = [
    "https://en.wikipedia.org/wiki/Fruit",
    "https://www.britannica.com/science/fruit-plant-reproductive-body",
]

# Validate URLs
valid_urls = [url for url in urls if is_valid_url(url)]
if not valid_urls:
    raise ValueError("No valid or accessible URLs found. Please check the URLs.")

# Load data
loader = UnstructuredURLLoader(urls=valid_urls)
data = loader.load()

# Split data into chunks
text_splitter = RecursiveCharacterTextSplitter(
    separators=['\n\n','\n', '.', ','],
    chunk_size=1000
)
docs = text_splitter.split_documents(data)

# Create embeddings
texts = [doc.page_content for doc in docs]

embeddings = embedding_model.embed_documents(texts)
# Convert embeddings to NumPy array (ensure it's 2D)
embeddings = np.array(embeddings)

# Create the FAISS index
embedding_dim = embeddings.shape[1]
index = faiss.IndexFlatL2(embedding_dim)

# Add embeddings to the FAISS index
index.add(embeddings)

# Create FAISS vector store using LangChain's FAISS class
faiss_index = FAISS.from_documents(docs, embedding_model)

print("FAISS index created successfully!")


```

```python
# Example query
query = "What is a fruit?"
k = 1 # Number of top results to retrieve

# Search the FAISS index
top_k_docs = faiss_index.similarity_search(query, k=k)
print("\nTop documents:")
for i, doc in enumerate(top_k_docs, 1):
    print(f"Document {i}:\n{doc.page_content}\n")
```
